```python

def calculate(operation, v1, v2):
    if operation == '+':
        return v1 + v2
    elif operation == '-':
        return v1 - v2
    elif operation == '*':
        return v1 * v2
    elif operation == '/':
        if v2 == 0:
            return 'Ошибкаж'
        result = v1 / v2
        return result
    else:
        return 'Ошибка'


def run_calculator():
    oper = input('Введите операцию')
    v_1 = int(input('первое число: '))
    v_2 = int(input('Второе число: '))
    res = calculate(oper, v_1, v_2)
    print('Рез', res)

run_calculator()
```

```python
# Подсчёт определённых символов в предложении

def count_chars(sentence, chars_dict):
    for char in sentence:
        if char in chars_dict:
            chars_dict[char] += 1
    return chars_dict


def run_count_chars():
    sentence = input('Введите предложение: ')
    chars = input('символы для подсчёта (без пробелов, например abc): ')

    # создаём словарь, где ключ - символ, а значение - количество
    chars_dict = {}
    for char in chars:
        chars_dict[char] = 0

    result = count_chars(sentence, chars_dict)

    for char in result:
        print(f"Символ '{char}' встречается {result[char]} раз(а)")

run_count_chars()
```

```python
# Количество символов из диапазона m - z

def count_m_to_z(text):
    count = 0
    for char in text:
        if char >= 'm' and char <= 'z':
            count += 1
    return count


def run_count_m_to_z():
    text = input('Введите строку: ')
    print('Количество символов от m до z:', count_m_to_z(text))

run_count_m_to_z()
```
