```python
# Задание 1. Будильник (логические операторы)
def need_alarm(employed, vacation):
    return employed and not vacation

employed = bool(input("Работаешь?: "))
vacation = bool(input("В отпуске?: "))

print(need_alarm(employed, vacation))

# Задание 2. Калькулятор через operation

def calculate(operation, value_1, value_2):
    if operation == '+':
        return value_1 + value_2
    elif operation == '-':
        return value_1 - value_2
    elif operation == '*':
        return value_1 * value_2
    elif operation == '/':
        # Проверка деления на ноль
        if value_2 == 0:
            return "Ошибка: деление на ноль"
        return value_1 / value_2
    else:
        return "Неизвестная операция"

oper = str(input("Оператор: "))
v_1 = int(input("Число 1: "))
v_2 = int(input("Число 2: "))
print(calculate(oper, v_1, v_2))

# Задание 3. Сумма чисел между двумя включительно

def sum_between(a, b):
    # Навсякий, если числа наоборот (5, 2)
    start = min(a, b)
    end = max(a, b)

    total = 0
    for number in range(start, end + 1):
        total += number
    return total

num_1 = int(input("Первое число: "))
num_2 = int(input("Второе число: "))

print(sum_between(num_1, num_2))

# Задание 4. Месяц → квартал и сезон

def month_info(month):
    if month < 1 or month > 12:
        return "Некорректный номер месяца"
    quarter = (month - 1) // 3 + 1

    if month in (12, 1, 2):
        season = "Зима"
    elif month in (3, 4, 5):
        season = "Весна"
    elif month in (6, 7, 8):
        season = "Лето"
    else:
        season = "Осень"

    return f"Месяц {month}: {quarter}-й квартал, сезон — {season}"

num_month = int(input("номер месяца: "))
print(month_info(num_month))

# Задание 5. Средний балл и буквенная оценка

def get_average_and_grade():
    grades = []

    # Бесконечный ввод до -1
    while True:
        value = int(input("Введите оценку (или -1 для завершения): "))
        if value == -1:
            break
        grades.append(value)

    if len(grades) == 0:
        return "Оценок не было введено"

    average = sum(grades) / len(grades)

    if average >= 90:
        letter = 'A'
    elif average >= 80:
        letter = 'B'
    elif average >= 70:
        letter = 'C'
    elif average >= 60:
        letter = 'D'
    else:
        letter = 'F'

    return f"Средний балл: {average}, оценка: {letter}"


print(get_average_and_grade())
```
