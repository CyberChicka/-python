```python
class Student:
    def __init__(self, name, group, tel_number):
        self.name = name
        self.group = group
        self.grades = []  # список оценок, сначала пустой
        self.__tel_number = tel_number  # приватный атрибут

    def add_grade(self, grade):
        # добавляем оценку в список
        self.grades.append(grade)

    def get_average(self):
        # если оценок нет, то среднего балла тоже нет
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def info(self):
        print("Имя:", self.name)
        print("Группа:", self.group)
        print("Телефон:", self.__tel_number)
        print("Оценки:", self.grades)
        print("Средний балл:", round(self.get_average(), 2))
        print()
```

```python
s_1 = Student("Анна", "ИС-21", "+7 900 111-22-33")
s_2 = Student("Иван", "ИС-21", "+7 900 444-55-66")
s_3 = Student("Мария", "ИС-22", "+7 900 777-88-99")
```

```python
s_1.add_grade(5)
s_1.add_grade(4)
s_1.add_grade(5)

s_2.add_grade(3)
s_2.add_grade(4)
s_2.add_grade(4)
s_2.add_grade(5)

s_3.add_grade(5)
s_3.add_grade(5)
s_3.add_grade(4)
```

```python
s_1.info()
s_2.info()
s_3.info()
```

```python
print("Средний балл Анны:", round(s_1.get_average(), 2))
print("Средний балл Ивана:", round(s_2.get_average(), 2))
print("Средний балл Марии:", round(s_3.get_average(), 2))
```
