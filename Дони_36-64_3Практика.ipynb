```python
# Задание 1. Поиск строки в списке
strings = ["машина", "дела", "телефон", "что-то", "ещё"]

search = input("Введите строку для поиска: ")

if search in strings:
    index = strings.index(search)
    print(f"Строка '{search}' найдена на позиции {index}")
else:
    print(f"Строки '{search}' нет в списке")

# Задание 2. Удаление дубликатов из списка

numbers = [1, 2, 2, 3, 4, 4, 4, 5, 1, 6]

unique = list(set(numbers))
print("Без дубликатов:", unique)

unique_ordered = []
for item in numbers:
    if item not in unique_ordered:
        unique_ordered.append(item)

print("С сохранением порядка:", unique_ordered)

# Задание 3. Подсчёт повторений через словарь

items = ["a", "b", "a", "c", "b", "a", "d"]
counts = {}

for item in items:
    if item in counts:
        counts[item] += 1
    else:
        counts[item] = 1

for key, value in counts.items():
    print(f"'{key}' встречается {value} раз")

# Задание 4. Удалить минимальное число, сохранив порядок
num = [5, 3, 8, 1, 9, 1, 4]
minimum = min(num)
num.remove(minimum)
print("Минимум:", minimum)
print("Список после удаления:", num)
```
