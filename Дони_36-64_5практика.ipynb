```python
import json

# читаем файл
try:
    f = open("students.json", "r", encoding="utf-8")
    students = json.load(f)
    f.close()
except FileNotFoundError:
    print("Ошибка: файл students.json не найден")
    exit()
except json.JSONDecodeError:
    print("Ошибка: в файле students.json неправильный формат данных")
    exit()

# проверяем что в файле именно список
if type(students) != list:
    print("Ошибка: в файле должен быть список студентов")
    exit()

report = []        # результат по каждому студенту
all_grades = []    # оценки всех студентов для общего среднего

for student in students:
    # если студент - не словарь, то пропускаем
    if type(student) != dict:
        print("Ошибка: студент должен быть словарем, пропускаем:", student)
        continue

    # проверка ключа name
    try:
        name = student["name"]
    except KeyError:
        print("Ошибка: у студента нет ключа name, пропускаем:", student)
        continue

    # проверка ключа grades
    try:
        grades = student["grades"]
    except KeyError:
        print("Ошибка: у студента", name, "нет ключа grades, пропускаем")
        continue

    # проверка типов данных
    if type(name) != str:
        print("Ошибка: имя должно быть строкой, пропускаем:", name)
        continue

    if type(grades) != list:
        print("Ошибка: оценки у", name, "должны быть списком, пропускаем")
        continue

    # проверяем что все оценки - числа
    grades_ok = True
    for g in grades:
        if type(g) != int and type(g) != float:
            grades_ok = False
    if grades_ok == False:
        print("Ошибка: у", name, "в оценках есть не числа, пропускаем")
        continue

    # если оценок нет, то делить на ноль нельзя
    try:
        avg = sum(grades) / len(grades)
    except ZeroDivisionError:
        print("Ошибка: у", name, "нет оценок, пропускаем")
        continue

    avg = round(avg, 2)
    print(name, "- средний балл:", avg)

    report.append({"name": name, "average": avg})
    all_grades = all_grades + grades

# общий средний балл
if len(all_grades) > 0:
    total_avg = round(sum(all_grades) / len(all_grades), 2)
else:
    total_avg = None
print("Общий средний балл:", total_avg)

# собираем итоговый отчет
result = {
    "students": report,
    "total_average": total_avg
}

# сохраняем в report.json
try:
    f = open("report.json", "w", encoding="utf-8")
    json.dump(result, f, ensure_ascii=False, indent=2)
    f.close()
    print("Отчет сохранен в report.json")
except OSError:
    print("Ошибка: не получилось записать файл report.json")
```
